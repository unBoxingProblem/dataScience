```python
import pandas as pd
import requests
import os
from tqdm import tqdm # Provides a nice progress bar!

# --- IMPORTANT: CHANGE THESE TWO VARIABLES ---
# 1. Replace with the path to your file. Find this path by clicking the "copy file path"
#    button next to your uploaded file in the "Input" section on the right.
file_path = '/kaggle/input/amazon/train.csv' 

# 2. Replace with the exact name of the column containing the image URLs.
url_column = 'image_link'
# ---------------------------------------------


# Create a directory to save the images
output_dir = '/kaggle/working/images'
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
    print(f"Created directory: {output_dir}")

# Read the file with the links
try:
    df = pd.read_csv(file_path)
    print(f"Successfully loaded {file_path}. Found {len(df)} rows.")
except FileNotFoundError:
    print(f"ERROR: File not found at {file_path}. Please check the file path.")
    df = None

if df is not None:
    # Loop through each URL in the specified column and download the image
    print(f"\nStarting image download from column '{url_column}'...")

    for index, row in tqdm(df.iterrows(), total=df.shape[0]):
        url = row[url_column]
        
        # Check if URL is a valid string
        if not isinstance(url, str):
            print(f"Skipping row {index}: URL is not a valid string.")
            continue

        try:
            response = requests.get(url, timeout=10) # 10-second timeout
            
            # Check if the request was successful
            if response.status_code == 200:
                # Create a filename (e.g., image_0.jpg, image_1.png)
                # Tries to get the original extension, otherwise defaults to .jpg
                file_extension = os.path.splitext(url)[1]
                if not file_extension:
                    file_extension = '.jpg'
                
                file_name = f"image_{index}{file_extension}"
                image_path = os.path.join(output_dir, file_name)
                
                # Save the image
                with open(image_path, 'wb') as f:
                    f.write(response.content)
            else:
                print(f"Skipping row {index}: Failed to download {url} with status code {response.status_code}")

        except requests.exceptions.RequestException as e:
            print(f"Skipping row {index}: Error downloading {url}. Reason: {e}")

    print("\n✅ Download complete!")
```
